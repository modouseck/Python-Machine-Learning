```python
import requests
import pandas as pd

url = "https://fakestoreapi.com/products"
response = requests.get(url)
data = response.json()

# Convertir les données en DataFrame
df = pd.DataFrame(data)
df.head()
```

```python
# Garder seulement les colonnes utiles
df = df[['title', 'price', 'category', 'rating']]
df['rating'] = df['rating'].apply(lambda x: x['rate'])

df.head()
```

```python
import plotly.express as px

fig = px.box(df, x='category', y='price', title='Distribution des prix par catégorie de produits')
fig.show()
```

```python
moyennes = df.groupby('category')['rating'].mean().reset_index()

fig2 = px.bar(moyennes, x='category', y='rating', title='Note moyenne par catégorie')
fig2.show()
```
